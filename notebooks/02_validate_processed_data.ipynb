# Part C: Check the processed files

This notebook compares the original files with the saved files.

**Raw** means original. **Processed** means the copy saved by your Python script.

Your current script does not remove rows or fill missing values. We therefore expect the tables to contain the same information. Customer ZIP-code prefixes should keep their leading zeros.

Run one cell at a time with **Shift + Enter**, starting at the top. Read the explanation and output before moving on. A result of `True` means the comparison matched. `False` means you need to investigate.

Open this notebook from the project's `notebooks` folder. The paths beginning with `../` go up to the project folder.

## 1. Load the tools

`pandas` reads and checks tables. We use the short name `pd`.

`Path` lets us read a file's original contents when checking that it stays unchanged.

```python
import pandas as pd
from pathlib import Path
```

## 2. Remember the original files before processing

These variables hold the exact file contents before the script runs. `read_bytes()` reads the file as stored on disk.

We will compare these contents after processing. This checks this run only; it cannot tell us whether a raw file was changed earlier.

```python
orders_file_before = Path("../data/raw/olist_orders_dataset.csv").read_bytes()
customers_file_before = Path("../data/raw/olist_customers_dataset.csv").read_bytes()
order_items_file_before = Path("../data/raw/olist_order_items_dataset.csv").read_bytes()
products_file_before = Path("../data/raw/olist_products_dataset.csv").read_bytes()
order_reviews_file_before = Path("../data/raw/olist_order_reviews_dataset.csv").read_bytes()
```

## 3. Run the processing script

`%run` is a notebook command that runs a Python file.

**This cell replaces the five saved CSVs in `data/processed/` with fresh outputs.** The script should leave the raw files unchanged.

```python
%run ../src/process_olist_data.py
```

## 4. Did the raw files stay unchanged?

`==` means "is equal to". Each line compares a raw file now with the contents remembered before the script ran.

Every result should be `True`. A `False` means that raw file changed during this run.

```python
print("orders unchanged:", orders_file_before == Path("../data/raw/olist_orders_dataset.csv").read_bytes())
print("customers unchanged:", customers_file_before == Path("../data/raw/olist_customers_dataset.csv").read_bytes())
print("order_items unchanged:", order_items_file_before == Path("../data/raw/olist_order_items_dataset.csv").read_bytes())
print("products unchanged:", products_file_before == Path("../data/raw/olist_products_dataset.csv").read_bytes())
print("order_reviews unchanged:", order_reviews_file_before == Path("../data/raw/olist_order_reviews_dataset.csv").read_bytes())
```

## Check the orders table

### Read both files

`orders_raw` holds the original table. `orders_saved` holds the saved table.

If this cell finishes without an error, Pandas could read both files.

```python
orders_raw = pd.read_csv("../data/raw/olist_orders_dataset.csv")
orders_saved = pd.read_csv("../data/processed/orders.csv")
print("Both files loaded successfully.")
```

### Check the size

`shape` shows `(rows, columns)`. The sizes should match because the script does not remove rows or columns.

```python
print("Raw size:", orders_raw.shape)
print("Saved size:", orders_saved.shape)
print("Same size:", orders_raw.shape == orders_saved.shape)
```

### Check the column names

`columns` lists the column names. `equals()` checks whether both lists match, including their order. Expect `True`.

```python
print("Raw columns:", orders_raw.columns.tolist())
print("Saved columns:", orders_saved.columns.tolist())
print("Same columns:", orders_raw.columns.equals(orders_saved.columns))
```

### Check the data types

`dtypes` shows how Pandas stores each column. Text is shown as `str`, `string`, or `object`; whole numbers as `int64`; decimal numbers as `float64`.

The columns should have the same types in both tables. Matching types does not prove they suit every task: date text would need conversion before date calculations.

```python
print("Raw types:")
print(orders_raw.dtypes)
print("Saved types:")
print(orders_saved.dtypes)
print("Same types:", orders_raw.dtypes.equals(orders_saved.dtypes))
```

### Check missing values

`isna().sum()` counts missing values in each column. The counts should stay the same because the script does not fill or remove missing values.

```python
print("Raw missing values:")
print(orders_raw.isna().sum())
print("Saved missing values:")
print(orders_saved.isna().sum())
print("Same missing counts:", orders_raw.isna().sum().equals(orders_saved.isna().sum()))
```

### Check duplicate rows

`duplicated().sum()` counts repeated complete rows. Repeated IDs alone are a different check.

The counts should stay the same. We have not decided to remove duplicates from these tables.

```python
print("Raw duplicates:", orders_raw.duplicated().sum())
print("Saved duplicates:", orders_saved.duplicated().sum())
print("Same duplicate count:", orders_raw.duplicated().sum() == orders_saved.duplicated().sum())
```

### Check categories and distinct values

`nunique()` counts different non-missing values in each column. These counts should match.

We also inspect `order_status`. `value_counts(dropna=False)` counts each value, including missing values. The comparison checks the complete counts; the next cell displays at most ten values.

```python
print("Same distinct counts:", orders_raw.nunique().equals(orders_saved.nunique()))
orders_raw_counts = orders_raw["order_status"].value_counts(dropna=False)
orders_saved_counts = orders_saved["order_status"].value_counts(dropna=False)
print("Same category counts:", orders_raw_counts.equals(orders_saved_counts))
```

```python
orders_saved_counts.head(10)
```

### Look at sample records

The next two cells display the first five rows of each version separately. Compare them: do their IDs, values, and missing entries match?

A sample helps you understand the data but does not check every row.

```python
orders_raw.head()
```

```python
orders_saved.head()
```

### Compare the whole table

The following check compares every row and column. Tiny decimal differences caused by saving and rereading CSV files are allowed; text must match.

`pd.testing.assert_frame_equal()` raises an error if the tables differ unexpectedly. If it passes, the success message appears.

```python
pd.testing.assert_frame_equal(orders_raw, orders_saved, check_exact=False, rtol=0.000000000001, atol=0.000000000001)
print("Passed: all table values match within the small decimal tolerance.")
```
